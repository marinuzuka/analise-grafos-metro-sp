# Análise de Grafos da Rede Metroferroviária de São Paulo

Este notebook apresenta a modelagem e construção de grafos referentes aos períodos de 1974, 1979 e 1996 da evolução da rede metroferroviária da Região Metropolitana de São Paulo.

As estações são representadas como vértices e as conexões entre estações como arestas, utilizando Python e a biblioteca NetworkX.

Os grafos construídos foram posteriormente exportados no formato GEXF para visualização e análise no Gephi.

```python
# funções
def criar_grafo():

    G = nx.Graph()

    return G


def adicionar_linha(grafo, linha):

    grafo.add_nodes_from(linha["estacoes"])

    grafo.add_edges_from(linha["arestas"])


def desenhar_grafo(grafo, posicoes, linhas, titulo,
                   integracoes=None,
                   tamanho_no=700,
                   tamanho_integracao=950):

    plt.figure(figsize=(20,16))

def desenhar_relatorio(grafo, posicoes, linha_estacao, titulo):

    plt.figure(figsize=(18, 14))

    cores = []

    for no in grafo.nodes():
        linha = linha_estacao.get(no, "Azul")
        cores.append(CORES[linha])

    nx.draw_networkx_nodes(
        grafo,
        posicoes,
        node_color=cores,
        node_size=180,
        edgecolors="black",
        linewidths=0.5
    )

    plt.title(
        titulo,
        fontsize=18,
        fontweight="bold"
    )

    plt.axis("off")

    plt.savefig(
        titulo.replace(" ", "_") + ".png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
```

```python
def calcular_metricas(grafo):

    return {
        "Estações": grafo.number_of_nodes(),
        "Conexões": grafo.number_of_edges(),
        "Densidade": round(nx.density(grafo),4),
        "Grau médio": round(
            sum(dict(grafo.degree()).values()) /
            grafo.number_of_nodes(),
            2
        ),
        "Componentes": nx.number_connected_components(grafo),
        "Diâmetro": nx.diameter(grafo) if nx.is_connected(grafo) else "-",
        "Raio": nx.radius(grafo) if nx.is_connected(grafo) else "-"
    }
```

```python
metricas = pd.DataFrame({

    "1974": calcular_metricas(G1974),

    "1979": calcular_metricas(G1979),

    "1996": calcular_metricas(G1996)

})

metricas
```

```python
def adicionar_atributos_arestas(grafo, arestas, nome_linha):

    for u, v in arestas:

        if grafo.has_edge(u, v):

            grafo[u][v]["linha"] = nome_linha
            grafo[u][v]["cor"] = CORES[nome_linha]
```

```python
def adicionar_posicoes(grafo, posicoes):

    for no, (x, y) in posicoes.items():

        if no in grafo.nodes:

            grafo.nodes[no]["x"] = float(x)
            grafo.nodes[no]["y"] = float(y)
```
